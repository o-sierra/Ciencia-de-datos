#● Lección 3: CORRELACIÓN

○ 📌 Objetivo: Detectar y cuantificar relaciones entre variables numéricas

○ 📍Tareas a desarrollar:


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

■ Construir scatterplots y matrices de correlación


```python
# Representación gráfica (Scartterplots)

variables_predictoras = ['Visitas','Ticket_Promedio', 'Monto_Total', 'Devoluciones']

fig, axes = plt.subplots(2, 2, figsize=(18, 5))
axes = axes.flatten()

for i, var in enumerate(variables_predictoras):
    sns.regplot(ax=axes[i], x=df[var], y=df['Compras'], scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
    axes[i].set_title(f'{var} vs Compras')

plt.tight_layout()
plt.show()
```

■ Calcular el coeficiente de Pearson (R)



```python
# Correlación de las variables númericas
numeric_df = df.select_dtypes(include=['number'])
correlaciones = numeric_df.corr(method='pearson') #method='pearson' lineal
correlaciones
```

```python
# visualizar el heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlaciones, annot=True, cmap='viridis', fmt=".2f", linewidths=0.5)
plt.title('Matriz de Correlación de Pearson')
plt.show()
```

# ● Lección 4: REGRESIONES LINEALES

○ 📌 Objetivo: Implementar un modelo de regresión simple y múltiple para explicar variables dependientes.

○ 📍Tareas a desarrollar:

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
X = df[['Monto_Total']]
y = df['Compras']
X

X_cons=sm.add_constant(X)
X_cons.head()
```

```python
#Crear el modelo de regresión lineal basado en OLS (Ordinary Least Squares)

modelo=sm.OLS(y,X_cons).fit()
```

```python
print(modelo.summary())
```

■ Calcular R² , MSE, MAE


```python
#MAE (error absoluto Medio)
mae=np.mean(np.abs(modelo.resid))
mae
```

```python
#MSE (Mean Squared Error)Error cuadrático promedio
mse=np.mean(modelo.resid**2)
mse
```

```python
#RMSE(raiz del error cuadrático promedio)
rmse=np.sqrt(mse)
rmse
```

```python
#Generar predicciones
y_pred=modelo.predict(X_cons)
y_pred
```
